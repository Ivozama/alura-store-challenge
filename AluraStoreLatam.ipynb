## Objetivo

Analizar el desempeño de cuatro tiendas para determinar cuál presenta menor rendimiento y debería ser vendida.

### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Facturación total por tienda
facturacion_t1 = tienda["Precio"].sum()
facturacion_t2 = tienda2["Precio"].sum()
facturacion_t3 = tienda3["Precio"].sum()
facturacion_t4 = tienda4["Precio"].sum()

print("Facturación Tienda 1:", facturacion_t1)
print("Facturación Tienda 2:", facturacion_t2)
print("Facturación Tienda 3:", facturacion_t3)
print("Facturación Tienda 4:", facturacion_t4)
```

# 2. Ventas por categoría

```python
categorias_t1 = tienda["Categoría del Producto"].value_counts()
categorias_t2 = tienda2["Categoría del Producto"].value_counts()
categorias_t3 = tienda3["Categoría del Producto"].value_counts()
categorias_t4 = tienda4["Categoría del Producto"].value_counts()

print("Tienda 1\n", categorias_t1)
print("Tienda 2\n", categorias_t2)
print("Tienda 3\n", categorias_t3)
print("Tienda 4\n", categorias_t4)
```

# 3. Calificación promedio de la tienda


```python
promedio_t1 = tienda["Calificación"].mean()
promedio_t2 = tienda2["Calificación"].mean()
promedio_t3 = tienda3["Calificación"].mean()
promedio_t4 = tienda4["Calificación"].mean()

print("Promedio Tienda 1:", promedio_t1)
print("Promedio Tienda 2:", promedio_t2)
print("Promedio Tienda 3:", promedio_t3)
print("Promedio Tienda 4:", promedio_t4)
```

# 4. Productos más y menos vendidos

```python
ventas_t1 = tienda["Producto"].value_counts()
ventas_t2 = tienda2["Producto"].value_counts()
ventas_t3 = tienda3["Producto"].value_counts()
ventas_t4 = tienda4["Producto"].value_counts()

print("Tienda 1 - Más vendido:", ventas_t1.idxmax())
print("Tienda 1 - Menos vendido:", ventas_t1.idxmin())

print("Tienda 2 - Más vendido:", ventas_t2.idxmax())
print("Tienda 2 - Menos vendido:", ventas_t2.idxmin())

print("Tienda 3 - Más vendido:", ventas_t3.idxmax())
print("Tienda 3 - Menos vendido:", ventas_t3.idxmin())

print("Tienda 4 - Más vendido:", ventas_t4.idxmax())
print("Tienda 4 - Menos vendido:", ventas_t4.idxmin())
```

# 5. Envío promedio por tienda

```python
envio_t1 = tienda["Costo de envío"].mean()
envio_t2 = tienda2["Costo de envío"].mean()
envio_t3 = tienda3["Costo de envío"].mean()
envio_t4 = tienda4["Costo de envío"].mean()

print("Costo promedio envío Tienda 1:", envio_t1)
print("Costo promedio envío Tienda 2:", envio_t2)
print("Costo promedio envío Tienda 3:", envio_t3)
print("Costo promedio envío Tienda 4:", envio_t4)
```

# 6. Graficos

```python
import matplotlib.pyplot as plt

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
facturaciones = [facturacion_t1, facturacion_t2, facturacion_t3, facturacion_t4]

plt.figure()
plt.bar(tiendas, facturaciones)
plt.title("Facturación Total por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos Totales")
plt.show()
```

```python
promedios = [promedio_t1, promedio_t2, promedio_t3, promedio_t4]

plt.figure()
plt.plot(tiendas, promedios, marker="o")
plt.title("Promedio de Calificación por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Calificación Promedio")
plt.show()
```

```python
envios = [envio_t1, envio_t2, envio_t3, envio_t4]

plt.figure()
plt.scatter(tiendas, envios)
plt.title("Costo Promedio de Envío por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Costo Promedio de Envío")
plt.show()
```
